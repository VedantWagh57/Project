```python
#calculate total sales 
df["TotalSales"] = df["Quantity"] * df["Price"]
```

```python
print(df[["Quantity", "Price", "TotalSales"]].head(10))
```

```python
print(df["TotalSales"].sum())
```

```python
product_sales = df.groupby("Description")["TotalSales"].sum()
```

```python
product_sales = product_sales.sort_values(ascending=False)
```

```python
print(product_sales)
```

```python
print(product_sales.head(10))
```

```python
import matplotlib.pyplot as plt

top_10 = product_sales.head(10)

plt.figure(figsize=(12, 6))

top_10.plot(kind="bar")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product Name")
plt.ylabel("Total Revenue")
plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
```

```python
#Monthly Sales Analysis.
monthly_sales = df.groupby(["Year", "Month"])["TotalSales"].sum()
```

```python
print(monthly_sales)
```

```python
#highest revenue month
print(monthly_sales.idxmax())
print(monthly_sales.max())
```

```python
#lowest revenue month
print(monthly_sales.idxmin())
print(monthly_sales.min())
```

```python
monthly_df = monthly_sales.reset_index()
```

```python
monthly_df["Year-Month"] = (
    monthly_df["Year"].astype(str) + "-" +
    monthly_df["Month"].astype(str).str.zfill(2)
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_df["Year-Month"],
    monthly_df["TotalSales"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Revenue")

plt.xticks(rotation=90)
plt.grid()

plt.tight_layout()
plt.show()
```

```python
#Task 15: Customer Sales Analysis
customer_sales = df.groupby("Customer ID")["TotalSales"].sum()
customer_sales = customer_sales.sort_values(ascending=False)
print(customer_sales.head(10))
print(customer_sales.head(10).sum())
```

```python
#Task 16: Top 10 Customers Visualization
top_customers = customer_sales.head(10)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

top_customers.plot(kind="barh")

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Customer ID")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
#Task 17: Country-wise Sales Analysis 🌍
country_sales = df.groupby("Country")["TotalSales"].sum()
country_sales = country_sales.sort_values(ascending=False)
print(country_sales.head(10))
print("="*30)
print(country_sales.idxmax())
print(country_sales.max())
```

```python
#Task 18: Country-wise Revenue Visualization 🌍
top_countries = country_sales.head(10)
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

top_countries.plot(kind="barh")

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Country")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
