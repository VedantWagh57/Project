```python
import pandas as pd

df = pd.read_csv(extract_path + r"\StudentsPerformance.csv")

print(df.head())
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print(df.dtypes)
```

```python
df["Total"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
)
```

```python
df["Average"] = df["Total"] / 3
```

```python
df["Result"] = np.where(
    df["Average"] >= 40,
    "Pass",
    "Fail"
)
```

```python
subjects = [
    "math score",
    "reading score",
    "writing score"
]

averages = df[subjects].mean()

averages.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.hist(df["math score"], bins=10)

plt.title("Maths Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
sns.barplot(
    data=df,
    x="gender",
    y="math score"
)

plt.title("Gender-wise Maths Performance")
plt.show()
```

```python
sns.barplot(
    data=df,
    x="test preparation course",
    y="Total"
)

plt.title("Test Preparation vs Total Marks")
plt.show()
```

```python
correlation = df[
    ["math score", "reading score", "writing score"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Subject Score Correlation")
plt.show()
```
